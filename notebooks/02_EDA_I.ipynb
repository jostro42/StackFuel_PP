#### Insights

- **Metadata**
    - only small number of entries with missing values in `summary`, publisher, and `releaseDate`
    - All columns coded as object. Think about potential transformations for ease of use further on
    - `platform` might be the only meta variable with a number of classes still usable with potential OHE later on. We'll need to think about how to best use the information hidden in `genre` and `publisher`
    - Year and month (and maybe even day) needs to be extracted and cyclically transformed from `releaseDate`
    - transformation of columns containing text (genre, summary, urls) to string?
- **Scores and counts**
    - a lot of missing data in critic and user variables. However, given the prior reformatting this is not too surprising. The original data set had 1 row per game, with addition information on scores and counts per platform hidden in a multitude of columns. By transforming into a game-platform format where we have multiple rows per game (1 per game platform combination), we automatically created rows with missing values. For some game-platform combinations, the score and review counts were unavailable
    - User score and counts were only available for the platform stated in the `section` variable of the original data set. Which is why we have ~60% missing values in the `user_[...]` variables. We went from 1 row to 3-4 rows per game on average, explaining the entry gap.
    - Ideally, missing values will be filled with subsequent scrapes.
    - Counts and critic scores need to be transformed to integers
    - vastly more user counts than critic counts...weighted counts needed for comparison!
    - massive discrepancy between min and max values in user review count numbers (e.g, 0 - 160k for `user_total_count`; gap a little smaller for different kinds of user reviews)


## Visualization

```python
# Check distributions of numerical variables
num_cols = num_vars.index

n_cols = min(3, len(num_cols))
n_rows = (len(num_cols) + n_cols - 1) // n_cols

fig, axes = plt.subplots(nrows = n_rows, ncols = n_cols, figsize = (13, n_rows * 3))
axes  = axes.flatten()

for idx, col in enumerate(num_cols):
    df[col].hist(bins = 30, ax = axes[idx], edgecolor = 'black', alpha = 0.6)
    axes[idx].set_title(f'{col}')
    axes[idx].grid(alpha = 0.3)

# do not display unused subplots
for idx in range(len(num_cols), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```

- highly skewed distributions of count variables, especially critic review counts!
    - Does not necessarily mean outliers...the discrepancy in player count between AAA and indie titles can be massive.
